```python
from datetime import date, datetime

TAXA_DIARIA = 0.025


def calcular_juros(valor, vencimento, hoje=None):
    if hoje is None:
        hoje = date.today()
    dias_atraso = (hoje - vencimento).days
    if dias_atraso <= 0:
        return 0, 0
    juros = valor * TAXA_DIARIA * dias_atraso
    return juros, dias_atraso


def formatar_real(valor):
    texto = f"{valor:,.2f}"
    texto = texto.replace(",", "X").replace(".", ",").replace("X", ".")
    return "R$ " + texto


def ler_valor():
    while True:
        texto = input("Valor (ex: 1500,00): ").strip()
        texto = texto.replace("R$", "").replace(" ", "")
        if "," in texto:
            texto = texto.replace(".", "").replace(",", ".")
        try:
            valor = float(texto)
        except ValueError:
            print("Valor inválido, tente de novo.")
            continue
        if valor <= 0:
            print("O valor precisa ser maior que zero.")
            continue
        return valor


def ler_data():
    while True:
        texto = input("Data de vencimento (dd/mm/aaaa): ").strip()
        try:
            return datetime.strptime(texto, "%d/%m/%Y").date()
        except ValueError:
            print("Data inválida, use o formato dd/mm/aaaa.")


def main():
    valor = ler_valor()
    vencimento = ler_data()

    juros, dias_atraso = calcular_juros(valor, vencimento)

    print()
    if dias_atraso == 0:
        print("Essa conta não está em atraso, então não tem juros.")
        return

    print(f"Dias de atraso: {dias_atraso}")
    print(f"Juros: {formatar_real(juros)}")
    print(f"Total a pagar: {formatar_real(valor + juros)}")


if __name__ == "__main__":
    main()
```
